```python
from huggingface_hub import snapshot_download

model_id = "Qwen/Qwen2.5-0.5B-Instruct"
local_folder = "Documents/Finetuning_few_shot/Qwen2.5-0.5B-Instruct"

print(f"Téléchargement de {model_id} vers {local_folder}...")

# L'argument crucial ici est local_dir_use_symlinks=False
# Cela force le téléchargement des vrais fichiers, pas juste des raccourcis cache.
snapshot_download(
    repo_id=model_id,
    local_dir=local_folder,
    local_dir_use_symlinks=False,
    ignore_patterns=["*.msgpack", "*.h5", "*.ot"] # On ignore les formats non-PyTorch pour gagner de la place
)

print("✅ Terminé ! Le dossier est prêt à être transféré.")
```
